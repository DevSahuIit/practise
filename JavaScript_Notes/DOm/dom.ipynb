# DOM (Document Object Model)

## 1. Definition of DOM

**DOM stands for Document Object Model.**

The DOM is a programming interface that represents an HTML document as a hierarchical tree of objects called nodes. The browser creates the DOM when it parses an HTML document, allowing JavaScript to access and modify the webpage's content, structure, and styles.

In simple words, the DOM converts an HTML document into a tree-like structure that JavaScript can manipulate.

### DOM stands for:

- **Document:** The webpage loaded in the browser.
- **Object:** Each HTML element is represented as an object that JavaScript can interact with.
- **Model:** The structured, tree-like representation of the webpage.

---

## 2. How HTML Is Converted into the DOM

### HTML Code

```html
<!DOCTYPE html>
<html>
<head>
    <title>My Website</title>
</head>
<body>
    <h1>Hello World</h1>
    <p>Welcome to JavaScript</p>
</body>
</html>
```

### DOM Tree Diagram

```text
                 document
                     |
                    html
                  /      \
               head      body
                |        /   \
              title     h1     p
                |        |     |
          "My Website"  "Hello World"
                       "Welcome to JavaScript"
```

### Explanation

1. The browser reads the HTML document.
2. It parses the HTML and creates a tree of nodes.
3. Each HTML element becomes an element node.
4. Text inside elements becomes text nodes.
5. JavaScript can access and modify these nodes through the DOM API.

**Important:** The DOM is a representation of the HTML document in memory. Changing the DOM does not necessarily change the original HTML source file.

---

## 3. Parent, Child and Sibling Relationships in DOM

Consider the following HTML code:

```html
<div>
    <h1>DOM Learning</h1>
    <p>Hello World</p>
</div>
```

### DOM Tree Diagram

```text
                  div
                 /   \
               h1     p
               |      |
        "DOM Learning" "Hello World"
```

### Relationships

- **Parent Node:** A node that directly contains another node.
  - `div` is the parent of `h1` and `p`.

- **Child Node:** A node directly contained inside another node.
  - `h1` and `p` are children of `div`.

- **Sibling Nodes:** Nodes that share the same parent.
  - `h1` and `p` are siblings.

- **Text Node:** A node containing text.
  - `"DOM Learning"` and `"Hello World"` are text nodes.

- **Root Node:** The topmost node in a tree.
  - In the complete browser DOM, `document` is the entry point, and `html` is the root element of the HTML document.

## 4. DOM Allows JavaScript to Change HTML

Consider this HTML:

```html
<h1 id="heading">Hello World</h1>
```

JavaScript can modify the heading using the DOM:

```javascript
document.getElementById("heading").textContent = "Welcome to JavaScript";
```

The browser will now display:

**Welcome to JavaScript**

The original HTML source remains unchanged, but the live DOM has been updated.

### Common DOM Operations

| Operation | Example |
|---|---|
| Select an element | `document.querySelector("h1")` |
| Select by ID | `document.getElementById("heading")` |
| Change text | `element.textContent = "Hello"` |
| Change HTML | `element.innerHTML = "<b>Hello</b>"` |
| Change CSS | `element.style.color = "red"` |
| Create an element | `document.createElement("p")` |
| Add an element | `parent.appendChild(element)` |
| Remove an element | `element.remove()` |

## 5. How the DOM Works

```text
        HTML Document
               |
               v
       Browser HTML Parser
               |
               v
         DOM Tree Created
               |
               v
      JavaScript Accesses DOM
               |
               v
      DOM Elements Modified
               |
               v
        Browser Updates UI
```

For example, when JavaScript changes the text of an element, the browser updates the rendered webpage to reflect that change.

## 6. Important Points to Remember

1. DOM stands for **Document Object Model**.
2. The DOM represents an HTML document as a tree of nodes and objects.
3. The browser creates the DOM by parsing HTML.
4. JavaScript uses DOM APIs to access and manipulate webpage elements.
5. Elements, attributes and text can be represented by different kinds of nodes.
6. Parent, child and sibling relationships help us navigate the DOM tree.
7. The DOM is a browser API, not a JavaScript language feature itself.
8. The DOM and the original HTML source are different: changes to the live DOM do not automatically rewrite the source HTML file.

# DOM Selectors, NodeList, HTMLCollection, and DOM Manipulation

## 1. What is the DOM?

The **Document Object Model (DOM)** is a programming interface that represents an HTML document as a tree of objects.

JavaScript uses the DOM to:
- Select HTML elements.
- Change text and HTML content.
- Modify CSS styles.
- Add, remove, or toggle CSS classes.
- Get and set HTML attributes.
- Create and remove elements.
- Respond to user events.

### Example HTML

```html
<!DOCTYPE html>
<html>
<head>
    <title>DOM Example</title>
</head>
<body>
    <h1 id="title">Hello World</h1>
    <p class="description">Learning JavaScript DOM</p>

    <ul>
        <li>HTML</li>
        <li>CSS</li>
        <li>JavaScript</li>
    </ul>
</body>
</html>
```

JavaScript can access these elements through the `document` object.

```javascript
console.log(document);
console.log(document.title);
console.log(document.body);
```

## 2. DOM Selectors

DOM selectors allow us to find HTML elements so that we can manipulate them.

### 2.1 `getElementById()`

Selects an element using its `id`.

```javascript
const titleElement = document.getElementById("title");

console.log(titleElement);
console.log(titleElement.textContent);
```

**Important points:**
- Returns a single element or `null` if no matching ID exists.
- IDs should be unique within an HTML document.
- Does not require `#` before the ID.

```javascript
document.getElementById("title");  // Correct
document.getElementById("#title"); // Incorrect
```

### 2.2 `getElementsByClassName()`

Selects all elements having a specified class name.

```javascript
const headings = document.getElementsByClassName("heading");

console.log(headings);
```

To access individual elements:

```javascript
console.log(headings[0]);
console.log(headings[1]);
```

**Important points:**
- Returns a live `HTMLCollection`.
- Uses the class name without the `.` symbol.
- Multiple elements can have the same class.

```javascript
document.getElementsByClassName("heading");  // Correct
document.getElementsByClassName(".heading"); // Incorrect
```

### 2.3 `getElementsByTagName()`

Selects all elements with a particular HTML tag.

```javascript
const paragraphs = document.getElementsByTagName("p");

console.log(paragraphs);
console.log(paragraphs[0]);
```

Examples:

```javascript
document.getElementsByTagName("h1");
document.getElementsByTagName("li");
document.getElementsByTagName("div");
```

**Important points:**
- Returns a live `HTMLCollection`.
- The selector is the tag name itself.
- Use indexing to access a particular element.

### 2.4 `querySelector()`

Selects the **first element** that matches a CSS selector.

```javascript
const title = document.querySelector("#title");
const heading = document.querySelector(".heading");
const paragraph = document.querySelector("p");
```

It supports standard CSS selector syntax.

| Selector | Meaning |
|---|---|
| `"#title"` | Element with ID `title` |
| `".heading"` | First element with class `heading` |
| `"p"` | First paragraph |
| `"ul li"` | First `li` inside a `ul` |
| `"div p"` | First paragraph inside a `div` |
| `"p:first-child"` | A `p` element that is the first child of its parent |
| `"input[type='password']"` | First password input |
| `"#container .item"` | First element with class `item` inside `#container` |

Example:

```javascript
const passwordInput =
    document.querySelector('input[type="password"]');

console.log(passwordInput);
```

This is the same type of selector shown in the tutorial screenshot.

If no matching element exists, `querySelector()` returns `null`.

### 2.5 `querySelectorAll()`

Selects **all elements** that match a CSS selector.

```javascript
const listItems = document.querySelectorAll("li");

console.log(listItems);
```

Other examples:

```javascript
const headings = document.querySelectorAll(".heading");

const paragraphs = document.querySelectorAll("p");

const passwordInputs =
    document.querySelectorAll('input[type="password"]');
```

**Important points:**
- Returns a `NodeList`.
- The `NodeList` returned by `querySelectorAll()` is static.
- It does not automatically update when elements are added or removed from the DOM.
- Supports `forEach()`.

## 3. Difference Between `HTMLCollection` and `NodeList`

This is an important interview and practical concept.

| Feature | HTMLCollection | NodeList from `querySelectorAll()` |
|---|---|---|
| Returned by | `getElementsByClassName()`, `getElementsByTagName()` | `querySelectorAll()` |
| Contents | Element nodes | Matching element nodes |
| Live or static | Live | Static |
| Indexing | Yes | Yes |
| `length` property | Yes | Yes |
| `forEach()` | Not directly supported as a standard method | Supported |
| Example | `collection[0]` | `nodeList[0]` |

### What does live mean?

A live collection automatically reflects relevant changes to the DOM.

```javascript
const liveItems = document.getElementsByTagName("li");

console.log(liveItems.length);
```

If a new `<li>` is added to the document, `liveItems.length` updates automatically.

### What does static mean?

A static `NodeList` represents the elements that matched when the selection was made.

```javascript
const staticItems = document.querySelectorAll("li");

console.log(staticItems.length);
```

If a new `<li>` is added later, `staticItems` does not automatically include it. You must run `querySelectorAll()` again to get an updated list.

**Remember:** Not every `NodeList` is static. For example, `childNodes` returns a live `NodeList`. The result of `querySelectorAll()` is static.

## 4. Selecting and Manipulating Multiple Elements

Consider this HTML:

```html
<ul>
    <li>HTML</li>
    <li>CSS</li>
    <li>JavaScript</li>
</ul>
```

### 4.1 Access an individual element

```javascript
const tempLiList = document.querySelectorAll("li");

console.log(tempLiList[0]);
console.log(tempLiList[1]);
console.log(tempLiList[2]);
```

### 4.2 Why does this produce an error?

```javascript
tempLiList.style.color = "green";
```

**Reason:** `tempLiList` is a `NodeList`, not an individual HTML element. It does not have the element's `style` property.

JavaScript tries to access a property that does not exist, causing an error.

### 4.3 Change the style of one element

```javascript
tempLiList[0].style.color = "green";
```

This changes only the first list item's text color.

### 4.4 Use `forEach()` to change every element

```javascript
tempLiList.forEach(function (item) {
    item.style.backgroundColor = "green";
});
```

Arrow-function version:

```javascript
tempLiList.forEach((item) => {
    item.style.backgroundColor = "green";
});
```

Both examples change the background color of every selected list item.

### 4.5 Use a `for...of` loop

```javascript
for (const item of tempLiList) {
    item.style.color = "green";
}
```

### 4.6 Use `Array.from()`

You can convert a collection into an array.

```javascript
const items = Array.from(
    document.getElementsByClassName("list-item")
);

items.forEach((item) => {
    item.style.color = "green";
});
```

This is useful when you want array methods on an `HTMLCollection`.

## 5. Changing CSS Styles with JavaScript

JavaScript can modify an element's inline CSS through the `style` property.

### 5.1 Change the background color

```javascript
const turnGreen = document.querySelector("p");

turnGreen.style.backgroundColor = "green";
```

### 5.2 Change text color

```javascript
turnGreen.style.color = "white";
```

### 5.3 Change font size

```javascript
turnGreen.style.fontSize = "20px";
```

### 5.4 Change multiple styles

```javascript
turnGreen.style.backgroundColor = "green";
turnGreen.style.color = "white";
turnGreen.style.fontSize = "20px";
turnGreen.style.padding = "10px";
turnGreen.style.borderRadius = "5px";
```

### 5.5 CSS property naming convention

CSS properties written with hyphens use camelCase in JavaScript.

| CSS property | JavaScript property |
|---|---|
| `background-color` | `backgroundColor` |
| `font-size` | `fontSize` |
| `border-radius` | `borderRadius` |
| `margin-top` | `marginTop` |
| `text-align` | `textAlign` |

### 5.6 Why is this incorrect?

```javascript
turnGreen.style.backgroundColor = "10px";
```

`backgroundColor` expects a valid CSS color, not a length.

To change the font size:

```javascript
turnGreen.style.fontSize = "10px";
```

To change the background color:

```javascript
turnGreen.style.backgroundColor = "green";
```

**Note:** The assignment of a valid CSS property value generally does not throw an error when the value is invalid; the browser may ignore the invalid value.

## 6. `innerHTML`, `innerText`, and `textContent`

These properties let you read or change the content of an HTML element.

Consider this HTML:

```html
<div id="example">
    Hello <b>Dev</b>
    <p>Learning DOM</p>
</div>
```

```javascript
const example = document.querySelector("#example");
```

### 6.1 `innerHTML`

`innerHTML` gets or sets the HTML markup inside an element.

```javascript
console.log(example.innerHTML);
```

Possible output:

```html
Hello <b>Dev</b>
<p>Learning DOM</p>
```

You can insert HTML elements using it:

```javascript
example.innerHTML = "<h2>Hello Dev</h2>";
```

The browser interprets the string as HTML and creates the corresponding elements.

You can also insert text and HTML together:

```javascript
example.innerHTML = "Welcome <b>Dev</b>";
```

The word `Dev` appears in bold.

**Important:** Assigning to `innerHTML` replaces the element's existing child content. Avoid inserting untrusted user input with `innerHTML`, because it can create cross-site scripting (XSS) vulnerabilities.

### 6.2 `innerText`

`innerText` gets or sets the rendered text of an element.

```javascript
console.log(example.innerText);
```

Possible output:

```text
Hello Dev
Learning DOM
```

It returns text as rendered by the browser, taking visibility and layout into account.

Setting it:

```javascript
example.innerText = "Hello JavaScript";
```

The element's content becomes plain text. Any previous child elements are replaced.

If you assign HTML-looking text:

```javascript
example.innerText = "<b>Hello</b>";
```

The browser displays:

```text
<b>Hello</b>
```

It does not create a bold element.

### 6.3 `textContent`

`textContent` gets or sets the text content of an element and its descendants, without interpreting the assigned value as HTML.

```javascript
console.log(example.textContent);
```

It includes text from descendant elements, even if some of that text is hidden by CSS.

Setting it:

```javascript
example.textContent = "Hello JavaScript";
```

The previous child elements are replaced with a text node.

If you assign:

```javascript
example.textContent = "<b>Hello</b>";
```

The browser displays the literal text:

```text
<b>Hello</b>
```

No bold element is created.

### 6.4 Main differences

| Feature | `innerHTML` | `innerText` | `textContent` |
|---|---|---|---|
| Reads content | HTML markup | Rendered text | Text from descendants |
| Preserves HTML tags when reading | Yes | No | No |
| Interprets assigned HTML | Yes | No | No |
| Includes hidden text when reading | Usually, through markup | Generally no | Yes |
| Affected by layout and visibility | Less directly | Yes | No |
| Replaces children when assigned | Yes | Yes | Yes |
| Suitable for untrusted plain text | Not directly | Yes | Yes |

### 6.5 Quick example

HTML:

```html
<div id="demo">
    Hello <span style="display: none;">Hidden</span>
    <b>World</b>
</div>
```

JavaScript:

```javascript
const demo = document.querySelector("#demo");

console.log(demo.innerHTML);
console.log(demo.innerText);
console.log(demo.textContent);
```

The outputs differ because:
- `innerHTML` returns the markup, including the hidden span.
- `innerText` returns rendered text and generally excludes the hidden text.
- `textContent` includes the text inside the hidden span.

**Remember:**
- Use `innerHTML` when you intentionally need to work with HTML markup.
- Use `innerText` when you need rendered, user-visible text.
- Use `textContent` for straightforward text access and safer text insertion.

## 7. Getting and Setting HTML Attributes

Attributes provide additional information about HTML elements.

Example:

```html
<input
    id="username"
    type="text"
    value="Dev"
    data-role="student"
>
```

### 7.1 `getAttribute()`

Reads the value of an HTML attribute.

```javascript
const input = document.querySelector("#username");

console.log(input.getAttribute("id"));
console.log(input.getAttribute("type"));
console.log(input.getAttribute("data-role"));
```

Output:

```text
username
text
student
```

If the attribute does not exist, `getAttribute()` returns `null`.

### 7.2 `setAttribute()`

Creates or updates an attribute.

```javascript
input.setAttribute("placeholder", "Enter your name");
input.setAttribute("data-role", "developer");
```

### 7.3 `removeAttribute()`

Removes an attribute.

```javascript
input.removeAttribute("placeholder");
```

### 7.4 `hasAttribute()`

Checks whether an attribute exists.

```javascript
console.log(input.hasAttribute("id"));
```

Returns `true` if the attribute exists and `false` otherwise.

### 7.5 Attribute versus DOM property

Some HTML attributes correspond to JavaScript properties.

```javascript
console.log(input.id);
console.log(input.type);
console.log(input.value);
```

There is an important distinction for form controls:

```javascript
input.getAttribute("value");
input.value;
```

The `value` attribute generally represents the initial/default value in the markup, whereas the `.value` property represents the current value of the input.

For example, if a user types into the input, `.value` changes even though the original `value` attribute may remain unchanged.

## 12. Complete Practical Example

This example combines selectors, text manipulation, styles, classes, and `forEach()`.

### HTML

```html
<!DOCTYPE html>
<html>
<head>
    <title>DOM Practice</title>

    <style>
        .highlight {
            background-color: yellow;
            color: black;
        }
    </style>
</head>
<body>
    <h1 id="title">DOM Tutorial</h1>

    <p class="description">Learning selectors</p>

    <ul>
        <li>HTML</li>
        <li>CSS</li>
        <li>JavaScript</li>
    </ul>

    <script>
        // Select by ID
        const title = document.getElementById("title");

        // Select the first paragraph
        const paragraph = document.querySelector("p");

        // Select all list items
        const items = document.querySelectorAll("li");

        // Change text
        title.textContent = "Learning DOM Manipulation";

        // Change inline styles
        paragraph.style.color = "blue";
        paragraph.style.fontSize = "20px";

        // Add a class
        paragraph.classList.add("highlight");

        // Change every list item's background color
        items.forEach((item) => {
            item.style.backgroundColor = "lightgreen";
        });

        // Display the number of list items
        console.log(items.length);
    </script>
</body>
</html>
```

## 13. Common Mistakes to Avoid

### Mistake 1: Using a CSS selector with `getElementById()`

Incorrect:

```javascript
document.getElementById("#title");
```

Correct:

```javascript
document.getElementById("title");
```

### Mistake 2: Applying styles to a `NodeList`

Incorrect:

```javascript
const items = document.querySelectorAll("li");

items.style.color = "green";
```

Correct:

```javascript
items.forEach((item) => {
    item.style.color = "green";
});
```

### Mistake 3: Forgetting that `querySelector()` returns one element

```javascript
const item = document.querySelector("li");

item.style.color = "green";
```

This changes only the first matching list item.

### Mistake 4: Confusing `innerHTML` with `textContent`

```javascript
element.innerHTML = "<b>Hello</b>";
```

Creates a bold element.

```javascript
element.textContent = "<b>Hello</b>";
```

Displays the literal characters `<b>Hello</b>`.

### Mistake 5: Using invalid CSS values

Incorrect:

```javascript
element.style.backgroundColor = "10px";
```

Correct:

```javascript
element.style.fontSize = "10px";
```

### Mistake 6: Selecting an element before it exists

If your JavaScript executes before the target HTML element has been parsed, a selector may return `null`.

One solution is to place the script near the end of the body:

```html
<body>
    <p id="message">Hello</p>

    <script>
        const message = document.querySelector("#message");
        message.style.color = "green";
    </script>
</body>
```

Alternatively, run your code after the DOM has loaded.

## 1. Parent-to-Child Navigation

### `parentElement.children`

## 2. Child-to-Parent & Sibling Traversal

Direct node relationships allow navigating the DOM tree without triggering additional expensive query operations (`document.querySelector` traverses the DOM tree, whereas property-based traversal is immediate via memory pointers).

| Property | Description |
| --- | --- |
| `element.parentElement` | Returns the direct parent element. |
| `element.nextElementSibling` | Returns the next sibling element (skips text/whitespace). |
| `element.previousElementSibling` | Returns the previous sibling element. |

```javascript
const child = document.querySelector('.child-class');

// Navigate to the direct parent
console.log(child.parentElement);

// Navigate to the adjacent sibling
console.log(child.nextElementSibling);

```

## 3. Elements vs. Nodes (`childNodes`)

* **`children` vs `childNodes`:**
* `children` selects **only HTML elements** (`<div>`, `<p>`, `<span>`, etc.).
* `childNodes` returns a **NodeList** containing **all nodes**, including:
* **Element nodes** (HTML tags)
* **Text nodes** (spaces, tabs, newlines/line breaks)
* **Comment nodes** (`<!-- comment -->`)





```javascript
console.log("Nodes: ", parent.childNodes);
// Line breaks and comments will appear as Text and Comment nodes

```

## 4. Creating and Styling DOM Elements in Memory

You can construct complete HTML elements dynamically in JavaScript memory before mounting them to the live DOM tree.

```javascript
// 1. Create element
const div = document.createElement('div');

// 2. Set identifiers
div.className = "main";
div.id = Math.floor(Math.random() * 10 + 1); // Generates an integer ID

// 3. Set attributes & inline styles
div.setAttribute("title", "generated title");
div.style.backgroundColor = "green";
div.style.padding = "12px";

```

---

## 5. Adding Text Content: `innerText` / `innerHTML` vs. `createTextNode`

There are two primary ways to populate an element's text content:

### Method A: `innerText` / `innerHTML`

Direct property assignment. While simple, `innerHTML` parses strings as HTML and forces an internal DOM re-parse.

```javascript
div.innerText = "Chai aur Code";

```

### Method B: `document.createTextNode()` + `appendChild()` *(Recommended for pure text)*

Creates a raw text node directly in memory without parsing overhead, then attaches it to the container.

```javascript
const addText = document.createTextNode("Chai aur Code");
div.appendChild(addText);

```

### 6. Attaching to Document Body

To make the created element visible on the page, mount it to an existing container (e.g., `document.body`):

```javascript
document.body.appendChild(div);

```

---

## Key Takeaways

1. **Performance:** Navigating relative relationships (`parentElement`, `nextElementSibling`, `children`) is faster than repeatedly running `document.querySelector`.
2. **Whitespace is a Node:** Always distinguish between `children` (elements only) and `childNodes` (elements + text + comments).
3. **Clean Text Insertion:** Prefer `document.createTextNode()` and `element.appendChild()` over `innerHTML` when inserting untrusted or plain text to prevent unnecessary re-parsing and potential XSS vectors.
