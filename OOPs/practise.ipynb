## Python `@property` Decorator

- `@property` is a decorator in Python that allows a method to be accessed like an attribute (variable), without using parentheses `()`.

### 1. Why is `@property` used?

- It allows us to access a method like a normal attribute.
- It provides controlled access to class attributes.
- It allows us to add logic when reading an attribute without changing how we access it.
- It helps us implement getters and setters.
- It makes code cleaner and easier to maintain.

### 2. Understanding the Code

```python
class User:
    def __init__(self, username, email, password):
        self.username = username
        self._email = email
        self.password = password

    @property
    def email(self):
        return self._email
````

- `self.username = username`: Stores the username.
- `self._email = email`: Stores the email address.
- `self.password = password`: Stores the password.
- `_email`: A single leading underscore indicates that the attribute is intended for internal use. It is a convention, not strict access control.
- `@property`: Allows the `email()` method to be accessed like an attribute.
- `return self._email`: Returns the stored email address.

### 3. How Does `@property` Work?

Without `@property`:

python
```
class User:
    def __init__(self, email):
        self._email = email

    def email(self):
        return self._email

user = User("abc@gmail.com")
print(user.email())
```

Output:

text
```
abc@gmail.com
```

- We must use parentheses `()` because `email` is a regular method.

With `@property`:

python
```
class User:
    def __init__(self, email):
        self._email = email

    @property
    def email(self):
        return self._email

user = User("abc@gmail.com")
print(user.email)
```

Output:

text
```
abc@gmail.com
```

- We do not need parentheses because `@property` allows us to access the method like an attribute.

### 4. What Is a Getter?

- A getter is a method used to retrieve or read an attribute's value.
- In Python, `@property` is commonly used to define a getter.
- In the example below, `email` is the getter.

python
```
@property
def email(self):
    return self._email
```

- When we write `user.email`, Python automatically calls the getter and returns `self._email`.

### 5. What Is a Setter?

- A setter is a method used to control how an attribute's value is changed.
- In Python, we use `@property_name.setter` to define a setter.
- A setter can validate data before storing it.

python
```
class User:
    def __init__(self, email):
        self._email = email

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, value):
        if "@" not in value:
            raise ValueError("Invalid email address")
        self._email = value
```

Usage:

python
```
user = User("abc@gmail.com")

print(user.email)
# abc@gmail.com

user.email = "hello"
# Raises ValueError: Invalid email address
```

- `@property` controls reading the email.
- `@email.setter` controls changing the email.
- `user.email` calls the getter.
- `user.email = "new@email.com"` calls the setter.
- If a setter is not defined, assigning a new value to `user.email` raises an `AttributeError`.
- A property without a setter is read-only through that property, but the underlying `_email` attribute can still be modified directly.

### 6. Important Points to Remember

- `@property` allows a method to behave like an attribute.
- A getter retrieves a value.
- A setter controls how a value is updated.
- The `@property` decorator does not automatically make an attribute private.
- A single leading underscore (`_email`) indicates intended internal use; it does not prevent access.
- Properties are useful for validation, calculated attributes, and controlling access to data.
- Properties help keep the same interface even when the implementation changes.

### 7. Quick Summary

| Syntax | Purpose |
| --- | --- |
| `@property` | Read a method like an attribute |
| `@email.setter` | Control how the email is updated |
| `user.email` | Read the email value |
| `user.email = "new@email.com"` | Update the email through the setter |
| `self._email` | Store the email using an internal-use naming convention |

In short: `@property` lets us access a method like an attribute while giving us the flexibility to add logic, validation, and controlled access when needed.

# Instance vs Static attributes


A static attributes (Some times called a class attributes) is an attribute that belongs to the class itse;lf, not any specific instance of the class

`````python
class User:
    user_count=0

    def __init__(self, username, email):
        self.username = username
        self.email = email
        User.user_count+=1

    def display(self):
        print(f"USername:{self.username} and Email: {self.email}")
    ## creating a protected method starts with a _
    def _is_valid_name(self,username):
        return username.length()>=8
    ## creating a private method
    def __name_good(self, username):
        return "a" in username

    ## this method remain same in all the objects formed by this class

    @staticmethod
    def is_valid_pass(password):
        return password.length()>=5
`````

`````python
user1 = User("Dev","Dev@1234" )
user2 = User("Deep", "deep@123en")
`````

`````python
print(User.user_count)
print(user1.user_count)
print(user2.user_count)
`````
