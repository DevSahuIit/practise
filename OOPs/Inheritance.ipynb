# Incheritance - is a fundamental concept in OOP that involves creating new classes (subclass or dervied classes) based on existing classes (superclasses or base classe)

## -A Car is -a Vehicle
## -A Bike is a Vehicle

```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start(self):
        print("Vehicle is Satrting")
    def stop(self):
        print("Vehicle is stopping")
```

```python
class Car(Vehicle): ## taht how we inherit
    def __init__(self, brand, model, year, noof_doors, no_wheels):
        super().__init__(brand, model, year)## baaps e kya kya mila hai isse
        self.noof_doors = noof_doors
        self.no_wheels = no_wheels
```

```python
class Bike(Vehicle):
    def __init__(self, brand, model, year, CCof_bike):
        super().__init__(brand, model, year)
        self.CCofbikes = CCof_bike
```

```python
car = Car("Ford", "Focus", 2008, 5, 4)
bike = Bike("Honda", "Scoopy", 2018, 2)
```
