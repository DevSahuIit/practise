## Abstraction - Reduce Complexity by hiding unnecessary details.

```python
class EmailService:
    def _connect(self):
        print("connecting to email server")
    def _authenticate(self):
        print("Authenticating....")

    def send_email(self):
        self._connect()
        self._authenticate
        print("Sending from Email server.....")
        self._disconnect()
    def _disconnect(self):
        print("Disconnecting from the server...")
```

```python
## we have pushed most of our function in 1 simple function thats how we are hiding complex codes
```
